```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("compressed_data.csv.gz")
```

```python
df.head()
```

```python
df.info()
```

checking missing values


```python
df.isnull().sum()
```

check,correct datatypes and remove column according to  required data

```python
df['last review']=pd.to_datetime(df['last review'],errors='coerce')
```

```python
df.fillna({'reviews per month': 0,
           'last review': df['last review'].min()},inplace=True)
```

```python
df.dropna(subset=['NAME','host name'],inplace=True)
```

```python
df.isnull().sum()
```

```python
df=df.drop(columns=['license','house_rules'],errors='ignore')
```

```python
df.head()
```

```python
df['price']=df['price'].replace('[\$,]', '',regex=True).astype(float)
df['service fee']=df['service fee'].replace('[\$,]', '',regex=True).astype(float)
```

```python
df.head()
```

remove duplicates


```python
df.drop_duplicates(inplace=True)
```

descriptive statistics

```python
df.describe()
```

visuallization


What is the distribution of listing Prices?

```python
plt.figure(figsize=(10,6))
sns.histplot(df['price'],bins=50,kde=True,color='red')
plt.title("Distribution of listing prices")
plt.xlabel("Price $")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="room type",data=df,color='blue')
plt.title("Distribution of room types")
plt.xlabel("Number of rooms")
plt.ylabel("Frequency")
plt.show()
```

how are listings distributed accross different neighborhoods?

```python
plt.figure(figsize=(12,8))
sns.countplot(y="neighbourhood group",data=df,color='blue',order=df['neighbourhood group'].value_counts().index)
plt.title("no. of listings by  neighborhood group")
plt.xlabel("count")
plt.ylabel("Neighbourhood group")
plt.show()

```

Relationship between price and roomtype

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='room type',y='price',hue='room type',data=df,palette='Set1')
plt.title("price vs. room types")
plt.xlabel("room type")
plt.ylabel("price $")
plt.show()
```
